```python
## CREATING DATAFRAME
import pandas as pd
sales = { "product":["Rice","Oil","Sugar","Banana","Bread","Beans"],
          "Quantity":[10,20,102,50,16,23,],
          "price":[500,1200,700,2300,4150,800]}
df =pd.DataFrame(sales)
print (df)
```

```python
## DATA AGGREGATION( groupby())
sales = {"Category":["Food","Food","Drink","Drink","Food","Drink"],
         "Products":["Rice","Beans","Juice","Water","Bread","Soda",],
         "Sales":[5000,2500,3500,2000,1500,4500]}
df =pd.DataFrame(sales)
print(df)
```

```python
print(df.groupby("Category")["Sales"].sum())
```

```python
print(df.groupby("Category")["Sales"].mean())
```

```python
print(df[df["Sales"]>1000])
```
